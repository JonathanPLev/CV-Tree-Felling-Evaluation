```python
from ultralytics import YOLO
```

```python
# Small-dataset baseline: start from a pretrained nano model and fine-tune conservatively.
model = YOLO('yolov8n.pt')
```

```python
# For limited data, prefer a lower learning rate, partial layer freezing,
# and moderate augmentation instead of the broad "augment=True" default.
# Use 640 so train/inference resolution matches a realistic iOS demo target.
results = model.train(
    data='forest_safety.yaml', #TODO: Create this file and the dataset directory before training
    epochs=150,
    imgsz=640,
    batch=8,
    patience=30,
    device=0,
    pretrained=True,
    freeze=10,
    optimizer='AdamW',
    lr0=0.001,
    lrf=0.01,
    weight_decay=0.0005,
    hsv_h=0.015,
    hsv_s=0.5,
    hsv_v=0.3,
    degrees=5,
    translate=0.05,
    scale=0.2,
    fliplr=0.5,
    mosaic=0.3,
    mixup=0.0,
    close_mosaic=10,
    cls=1.0,
    box=7.5,
    plots=True,
)
```

```python
# Export for iOS deployment after training completes.
# `nms=True` bakes NMS into the CoreML model for simpler app-side integration.
model.export(
    format='coreml',
    imgsz=640,
    nms=True,
    half=True,
)
```
